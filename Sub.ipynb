#1️⃣ Convert country to Uppercase


we standardize the country column by converting all values to uppercase?


```python
Sub['country'] = Sub['country'].str.upper()
```

This ensures consistency in text formatting and prevents issues caused by mixed case values

#2️⃣ Create Annual Income Column

we create a new column annual_income from monthly_income?

```python
Sub['annual_income'] = Sub['monthly_income'] * 12
```

This derives yearly income from monthly income, which may be more useful for financial analysis

#3️⃣ Create High Engagement Flag

we identify users with above-average engagement?

```python
Sub['high_engagement'] = Sub['time_spent_minutes'] > Sub['time_spent_minutes'].mean()
```

This creates a Boolean column identifying users who spend more time than average on the platform.

#4️⃣ Create Frequent Visitor Column

we classify users as frequent visitors based on website visits?

```python
Sub['frequent_visitor'] = Sub['website_visits'] > 10
```

#5️⃣ Rename Column

we rename time_spent_minutes to time_spent for simplicity?

```python
Sub = Sub.rename(columns={'time_spent_minutes': 'time_spent'})
```

#6️⃣ Convert device_type to Uppercase

we standardize the device_type column by converting values to uppercase?



```python
Sub['device_type'] = Sub['device_type'].str.upper()
```

This ensures uniform formatting and avoids inconsistencies in categorical data.

#7️⃣ Create Income per Age Ratio
we create a new column measuring income relative to age?

```python
Sub['income_per_age'] = Sub['monthly_income'] / Sub['age']
```

This derived feature may help analyze financial capacity adjusted for age.

#8️⃣ Create Returning Customer Flag
identify returning customers based on previous purchases?

```python
Sub['returning_customer'] = Sub['previous_purchases'] > 0
```

This categorizes users as new or returning customers, useful for behavioral analysis.

9️⃣ Remove Extra Spaces from country
clean the country column by removing leading/trailing spaces?

```python
Sub['country'] = Sub['country'].str.strip()
```

Removing whitespace prevents duplicate-like entries caused by hidden spaces.

🔟 Create Total Engagement Column
create a combined engagement metric using visits and time spent?

```python
Sub['total_engagement'] = Sub['website_visits'] + Sub['time_spent']
```

This creates a composite metric representing overall user engagement.
